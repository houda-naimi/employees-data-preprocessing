8. For all categorical attributes, replace missing values with the most frequent value using fillna() function.

```python
#8
categorical_columns = df.select_dtypes(include=['object', 'category'])

for column in categorical_columns.columns:
    most_frequent_value = df[column].mode()[0]
    df[column] = df[column].fillna(most_frequent_value)

print("\nDataset After Replacing Missing Values in Categorical Attributes:")
print(df.head())
```

9. Handle missing values NA, for the numerical attributes:

```python
#9
import numpy as np
from sklearn.linear_model import LinearRegression

# Remove the column having the most missing values.
column_with_most_na = df.isnull().sum().idxmax()
df = df.drop(columns=[column_with_most_na])
print(f"Column '{column_with_most_na}' has been removed.")

# Replace the missing values of the ‘statutory-holidays’ column by its maximum value.
if 'statutory-holidays' in df.columns:
    max_statutory_holidays = df['statutory-holidays'].max()
    df['statutory-holidays'] = df['statutory-holidays'].fillna(max_statutory_holidays)
    print("Missing values in 'statutory-holidays' replaced with maximum value.")

# Replace the missing values of the ‘working-hours’ column by its median value.
if 'working-hours' in df.columns:
    median_working_hours = df['working-hours'].median()
    df['working-hours'] = df['working-hours'].fillna(median_working_hours)
    print("Missing values in 'working-hours' replaced with median value.")

# Replace the missing values of the ‘delay-hours’ column
# by its mean value according to the ‘class’ attribute of the employee.
if 'delay-hours' in df.columns and 'class' in df.columns:
    class_grouped_mean = df.groupby('class')['delay-hours'].transform('mean')
    df['delay-hours'] = df['delay-hours'].fillna(class_grouped_mean)
    print("Missing values in 'delay-hours' replaced with mean grouped by 'class'.")

# Determine the average of 'delay-hours' by 'class'
if 'delay-hours' in df.columns and 'class' in df.columns:
    delay_hours_avg_by_class = df.groupby('class')['delay-hours'].mean()
    print("\nAverage of 'delay-hours' by 'class':")
    print(delay_hours_avg_by_class)

# Function to replace missing 'delay-hours' values according to 'class'
def remove_delay_hours_na(row):
    if pd.isnull(row['delay-hours']):
        return delay_hours_avg_by_class[row['class']]
    return row['delay-hours']

if 'delay-hours' in df.columns and 'class' in df.columns:
    df['delay-hours'] = df.apply(remove_delay_hours_na, axis=1)
    print("Applied function to replace missing 'delay-hours' values.")

# Replace the missing values of the ‘wage-increase-global’ column
#by fitting a linear regression model using other numerical columns in the dataset.

if 'wage-increase-global' in df.columns:
    df_num = df.select_dtypes(include=[np.number])

    traindf = df_num[df_num['wage-increase-global'].notnull()]
    testdf = df_num[df_num['wage-increase-global'].isnull()]

    X_train = traindf.drop(columns=['wage-increase-global'])
    y_train = traindf['wage-increase-global']
    X_test = testdf.drop(columns=['wage-increase-global'])

    lr = LinearRegression()
    lr.fit(X_train, y_train)

    y_pred = lr.predict(X_test)

    df.loc[df['wage-increase-global'].isnull(), 'wage-increase-global'] = y_pred
    print("Missing values in 'wage-increase-global' filled with predicted values.")
```

10. Create a function, named find_outliers_IQR(df), to find the outliers data in the ‘working-hours’ column based on the IQR statistical method. You can use quantile() function. Then, depending on the outlier’s value, replace it with the Q3 value (if it is greater than the maximum value) or replace it with the Q1 value (if it is less than the minimum value).

```python
#10
def find_outliers_IQR(df):
    Q1, Q3 = df['working-hours'].quantile([0.25, 0.75])

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df['working-hours'] = df['working-hours'].apply(
        lambda x: Q1 if x < lower_bound else (Q3 if x > upper_bound else x)
    )

    return df

print("Original DataFrame:")
print(df.head())

df = find_outliers_IQR(df)

print("\nDataFrame after handling outliers:")
print(df.head())
```
